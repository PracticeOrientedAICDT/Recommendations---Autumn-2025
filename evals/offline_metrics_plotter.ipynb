```python
import json
import os
import pandas as pd
import plotly.graph_objects as go
```

```python
base_path = "eval_results/"
recommenders_path = "overall_metrics_microsoft_recommenders.json"
MF_NMF_path = "overall_metrics_NMF_MF.json"
```

```python
# Compile data
with open(os.path.join(base_path, recommenders_path), 'r') as file:
    data = json.load(file)

with open(os.path.join(base_path, MF_NMF_path), 'r') as file:
    data["NMF"] = json.load(file)

overall_metrics_df = pd.DataFrame(data)
display(overall_metrics_df)
```

```python
plotting_df = overall_metrics_df.transpose().reset_index(names="model")
plotting_df
```

### Precision, recall and F1

```python
import seaborn as sns
import plotly.graph_objects as go
```

```python

colours = ["#ffb000",  "#dc267f",  "#648fff", "#fe6100", "#785ef0"]  # - IBM colourblind palette


fig = go.Figure(
    data=[
        go.Bar(name='Precision@K', x=plotting_df.model, y=plotting_df.precision_at_k, offsetgroup=1,
               marker_color="#648fff"),
        go.Bar(name='Recall@K', x=plotting_df.model, y=plotting_df.recall_at_k, offsetgroup=2, marker_color= "#dc267f"),
        go.Bar(name='F1@K', x=plotting_df.model, y=plotting_df.f1_at_k, offsetgroup=3, marker_color="#ffb000"),
    ],
    layout={
        #'title': {'text': ''},
        'xaxis': {'title': 'Model'},
        'yaxis': {'title': 'Mean of Precsion/Recall/F1 Score'}
    }
)

# Change the bar mode
fig.update_layout(barmode='group', height=400, width=800)
fig.show()
```
