```python
import json
import os
import pandas as pd
import plotly.graph_objects as go
```

```python
base_path = "eval_results/"
recommenders_path = "overall_metrics_microsoft_recommenders.json"
MF_NMF_path = "overall_metrics_NMF_MF.json"
BERT4Rec_path = "overall_metrics_bert4rec.json"
DiffRec_path = "overall_metrics_diffrec_recbole.json"
microsoft_benchmarks_path = "overall_metrics_microsoft_benchmark.json"
```

```python
# Compile data
with open(os.path.join(base_path, recommenders_path), 'r') as file:
    data = json.load(file)

with open(os.path.join(base_path, MF_NMF_path), 'r') as file:
    data["NMF"] = json.load(file)

with open(os.path.join(base_path, BERT4Rec_path), 'r') as file:
    data["BERT4Rec"] = json.load(file)

with open(os.path.join(base_path, DiffRec_path), 'r') as file:
    data["DiffRec"] = json.load(file)

overall_metrics_df = pd.DataFrame(data)
display(overall_metrics_df)
```

```python
plotting_df = overall_metrics_df.transpose().reset_index(names="model")
plotting_df
```

```python
with open(os.path.join(base_path, microsoft_benchmarks_path), 'r') as file:
    ms_benchmarks = json.load(file)
ms_benchmarks_df = pd.DataFrame(ms_benchmarks)
ms_benchmarks_df = ms_benchmarks_df.transpose().reset_index(names="model")
ms_benchmarks_df
```

### Precision, recall and F1

```python
colours = ["#ffb000",  "#dc267f",  "#648fff", "#fe6100", "#785ef0"]  # - IBM colourblind palette
```

```python

fig = go.Figure(
    data=[
        go.Bar(name='Precision@10', x=plotting_df.model, y=plotting_df.precision_at_k, offsetgroup=1,
               marker_color="#648fff"),
        go.Bar(name='Precision@10 (benchmark)', x=ms_benchmarks_df.model, y=ms_benchmarks_df.precision_at_k,
               offsetgroup=1, marker_color="#648fff", opacity=0.25),
        go.Bar(name='Recall@10', x=plotting_df.model, y=plotting_df.recall_at_k,
               offsetgroup=2, marker_color= "#dc267f"),
        go.Bar(name='Recall@10 (benchmark)', x=ms_benchmarks_df.model, y=ms_benchmarks_df.recall_at_k,
               offsetgroup=2, marker_color="#dc267f", opacity=0.25),
        go.Bar(name='F1@10', x=plotting_df.model, y=plotting_df.f1_at_k, offsetgroup=3, marker_color="#ffb000"),
        go.Bar(name='F1@10 (benchmark)', x=ms_benchmarks_df.model, y=ms_benchmarks_df.f1_at_k,
               offsetgroup=3, marker_color="#ffb000", opacity=0.25)
    ],
    layout={
        #'title': {'text': ''},
        'xaxis': {'title': 'Model'},
        'yaxis': {'title': 'Mean of Precsion/Recall/F1 Score @10'}
    }
)

# Change the bar mode
fig.update_xaxes(categoryorder="category ascending")
fig.update_layout(barmode='group', height=400, width=800)
fig.show()
```
